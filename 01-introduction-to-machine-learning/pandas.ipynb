# Important Pandas Library

Pandas is a Python library used for working with structured or tabular data.

In Machine Learning, Pandas is commonly used to load, inspect, clean, transform, and prepare datasets before training a model.


```python
import numpy as np
import pandas as pd
```

## DataFrame

A **DataFrame** is a two-dimensional table used to store structured data.

- Rows = Samples / Observations
- Columns = Features / Variables

In Machine Learning, a DataFrame is commonly used to inspect, clean, transform, and prepare a dataset before training a model.


```python
data = [
    ['Nissan', 'Stanza', None, 138, 4, 'MANUAL', 'sedan', 2000],
    ['Hyundai', 'Sonata', 2017, 180, 4, 'AUTOMATIC', 'Sedan', 27150],
    ['Toyota', 'Corolla', 2018, 132, 4, 'AUTOMATIC', None, 19500],
    ['Honda', 'Civic', 2020, 158, 4, 'AUTOMATIC', 'sedan', 22000],
    ['Ford', 'Mustang', None, 310, 4, 'AUTOMATIC', 'coupe', 35500],
    ['Chevrolet', 'Malibu', 2016, 160, 4, 'AUTOMATIC', 'sedan', 23500],
    ['Mazda', 'MX-5', 2021, 181, 4, 'MANUAL', 'convertible', 27000],
    ['BMW', '3 Series', 2020, 255, 4, 'AUTOMATIC', 'sedan', 41000],
    ['Mercedes-Benz', 'C-Class', 2019, 255, 4, 'AUTOMATIC', 'sedan', None],
    ['Subaru', 'Impreza', 2018, 152, 4, 'MANUAL', 'hatchback', 21000]
]

columns = [
    'Make',
    'Model',
    'Year',
    'Engine HP',
    'Engine Cylinders',
    'Transmission Type',
    'Vehicle_Style',
    'MSRP'
]

# Create a DataFrame from the data and column names
df = pd.DataFrame(data, columns=columns)

df
```

```python
df.head(n=2)
```

## Series

A **Series** is a one-dimensional data structure in Pandas and can usually be thought of as one column of a DataFrame.

In Machine Learning, a Series can represent a single Feature or a Target variable.


```python
df['Engine HP']  # Select one column as a Series
```

```python
df[['Make', 'Model', 'MSRP']]  # Select multiple columns
```

Selecting multiple columns is commonly used when choosing Features that will later form a **Feature Matrix**.


```python
df['id'] = range(1, len(df) + 1)  # Add an ID column based on the number of rows
df
```

```python
del df['id']  # Delete the ID column
df
```

## Index

An **Index** identifies the rows of a DataFrame.

It helps us locate, select, and manage individual Samples or Observations in a dataset.


```python
df.index
```

```python
df.index = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']
df
```

```python
df.iloc[[1, 2, 4]]
```

`iloc` means **Integer Location** and is used to select rows by their numerical position.


```python
df = df.reset_index(drop=True)
df
```

## Accessing Elements

Accessing elements means selecting specific Columns, Rows, or values from a DataFrame.

This is useful when exploring a dataset or selecting Features and Samples for further analysis.


```python
df['MSRP']  # Select one Column
```

```python
df[['Make', 'Model']]  # Select multiple Columns
```

```python
df.iloc[5]  # Select the sixth Row
```

## Element-wise Operations

Element-wise operations apply the same operation to every value in a Series or Column.

In Machine Learning, these operations are useful for transforming Features, creating new Features, or applying conditions to data.


```python
df['Engine HP'] * 2
```

```python
df['Year'] >= 2015
```

The comparison returns Boolean values:

- `True` = the condition is satisfied
- `False` = the condition is not satisfied

These Boolean values can later be used for Filtering.


## Filtering

Filtering is used to select only the Rows that satisfy a specific condition.

It is commonly used in Data Cleaning, Data Analysis, and creating subsets of a dataset.


```python
df[df['Make'] == 'Nissan']
```

```python
df[
    (df['Make'] == 'Hyundai') &
    (df['Year'] >= 2015)
]
```

When combining multiple conditions:

- `&` = AND
- `|` = OR


## String Operations

String Operations are used to clean and standardize text or Categorical Features.

This is important because values such as `Sedan`, `SEDAN`, and `sedan` may otherwise be treated as different categories.


```python
'machine learning zoomcamp'.replace(' ', '_')
```

```python
df['Vehicle_Style'].str.lower()
```

```python
df['Vehicle_Style'] = (
    df['Vehicle_Style']
    .str.replace(' ', '_')
    .str.lower()
)

df
```

- `.str.replace(' ', '_')` replaces spaces with underscores.
- `.str.lower()` converts text to lowercase.

Example:

`four door sedan` → `four_door_sedan`


## Summarizing Operations

Summarizing Operations help us understand the overall characteristics of a dataset.

They are an important part of **Data Understanding** and **Exploratory Data Analysis (EDA)** before training a Machine Learning model.


```python
df.describe().round(3)  # Round the summary statistics to 3 decimal places
```

`describe()` calculates summary statistics for Numerical Columns, such as:

- count
- mean
- standard deviation
- minimum
- quartiles
- maximum

These statistics help us understand the Scale, Distribution, and possible unusual values in Features.


```python
df.nunique()
```

`nunique()` counts the number of unique values in each Column.

It is especially useful for understanding Categorical Features and checking how many Categories are present.


## Missing Values

Missing Values are data points that are unavailable or not recorded.

In this dataset, `None` is used to represent missing information.

Checking Missing Values is important because many Machine Learning algorithms cannot work with missing data directly and may require the values to be handled before training.


```python
df.isnull()
```

`isnull()` checks every value in the DataFrame and returns:

- `True` = Missing Value
- `False` = Value is available


```python
df.isnull().sum()
```

`isnull().sum()` counts the number of Missing Values in each Column.

This helps identify which Features contain missing data and may require Data Cleaning before training a model.


```python
df.isnull().mean() * 100
```

`isnull().mean() * 100` calculates the percentage of Missing Values in each Column.


## Grouping

Grouping divides a dataset into Categories and then calculates a summary for each group.

It helps us explore Patterns and differences between groups within the dataset.


```python
df.groupby('Transmission Type').MSRP.max()
```

This groups the data by `Transmission Type` and calculates the maximum `MSRP` for each group.

Grouping is useful when exploring the relationship between a Categorical Feature and a Numerical Feature.


## Getting NumPy Arrays

Pandas is useful for preparing and managing datasets, while many Machine Learning algorithms work internally with NumPy Arrays or Matrices.

Therefore, Pandas data can be converted into NumPy format when needed.


```python
df.MSRP.values
```

```python
df['MSRP'].to_numpy()
```

`to_numpy()` converts a Pandas Series into a NumPy Array.

For example, a Target Column can later be converted into the Target Vector `y`.


```python
df.to_dict(orient='records')
```

`to_dict(orient='records')` converts a DataFrame into a List of Dictionaries.

Each Dictionary represents one Row or Record.

This format is useful when transferring data to other Python code, APIs, or application systems.
