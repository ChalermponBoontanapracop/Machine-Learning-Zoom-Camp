Q7. Sum of weights: What's the sum of all the elements of the result?

```python
df_asia = df[df["origin"] == "Asia"] # Select all the cars from Asia
X = df_asia[["vehicle_weight", "model_year"]] # Select columns vehicle_weight and model_year
X = X.head(7).to_numpy() #Select the first 7 values and Get the underlying NumPy array. Call it X.
print(X)
```

```python
XTX = X.T.dot(X) # Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Call the result XTX.
print(XTX)
```

```python
XTX_inv = np.linalg.inv(XTX) # Invert XTX.
print(XTX_inv)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200]) # Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
w = XTX_inv.dot(X.T).dot(y) # Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
print(w)
```

```python
w.sum() # Sum of all the elements of the result
```
